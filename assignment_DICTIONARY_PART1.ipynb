```python
# #2. Create a tuple of tuples, where each inner tuple contains a name and an age.
#     A. Convert this data into a dictionary where names are keys and ages are values.
#     b. Perform various dictionary operations like
#         1) adding a new entry, 2) updating an age and 3) deleting an entry.


people = (("anu", 20), ("achu", 22), ("appu", 15))
#print(people, type(people))
dict1 = dict(people)
print(dict1, "is the existing Dictionary")
while True:
  print()
  message = "Enter an option:\n"
  message += "1. 'ADD'- Adding new person\n"
  message += "2. 'UPDATE'- Update any of age\n"
  message += "3. 'DELETE'- Delete an entry\n"
  message += "4. 'EXIT'- Exit from the system"
  option = input(message).lower()
  print()

# stop the running
  if option in ["exit", "4", "e"]:
    print("bye!!")
    break

# add new key and value
  elif option in ["add", "1", "a"]:
    name = input("Enter the name :- ")
    age = int(input("Enter the age :- "))
    dict1[name] = age
    print(dict1, type(dict1))

# update value. If there is  no matching key, add new key and value
  elif option in ["update", "2", "u"]:
    name = input("Enter the name of the person whose AGE to be changed :- ")
    if name not in dict1:
      print(name.upper(), "is not in the Dictionary.")
      print("Do you wanna add he/she to the Dictionary?")
      decision = input("YES/NO").lower()
      if decision == "yes":
        age = int(input("Enter the age :- "))
        dict1[name] = age
      else:
        print("OK")
        continue
    else:
      age = int(input("Age to be replaced as :- "))
      dict1[name] = age
    print(dict1, type(dict1))

# deleting a key:value pair
  elif option in ["delete", "3", "d"]:
    name = input("Enter the name who need to be removed :- ")
    if name not in dict1:
      print(name.upper(), "is not in the existing dictionary.")
    else:
      print(name, "is removed from the Dictionary")
      dict1.pop(name)
      print(dict1)
  else:
    print("Enter a valid option!!!")
```
