```python
# 1. Write a program that converts all vowels (a, e, i, o, u) in a string to uppercase,
#    while keeping the other characters unchanged.

# Input: "hello world"
# Output: "hElLO wOrld

text = input("Enter input : ")
out = ""
for i in text:
  if i in "aeiouAEIOU":
    out += i.upper()
  else:
    out += i
print(out)
```

```python
# 2. Write a program that converts all consonants in a string to uppercase,
#    while leaving vowels and non-alphabetic characters unchanged.

#  Input: "hello world"
# Output: "HELLO wORID"

text = input("Enter input : ")
out = ""
for i in text:
  if i not in "aeiouAEIOU":
    out += i.upper()
  else:
    out += i
print(out)
```

```python
# 3. Write a program that converts a camelCase string into snake_case
#    (all lowercase,words separated by underscores).

#  Input: "camelCaseExample"
# Output: "camel_case_example

text = input("Enter camelcase string : ")
out = ""

for i in text:
  if i == i.upper():
    out += "_" + i.lower()
  else:
    out += i
print("snake_case = ", out)
```

```python
# 6. Write a function that inverts the case of letters at even indices
#    (make them uppercase if they're lowercase, and vice versa),
#    while leaving the letters at odd indices unchanged.
# Input: "hello"
# Output: "HeLlO"

text = input("Enter a text : ")
out = ""

for i in range(len(text)):
  if i%2 == 0:
    if text[i] == text[i].lower():
      out += text[i].upper()
    elif text[i] == text[i].upper():
      out += text[i].lower()
  else:
    out += text[i]
print(out)
```

```python
# 8. Count all letters, digits, and special symbols from a given string.

text = input("Enter a string : ")
alpcount = 0
numcount = 0
specount = 0

for i in text:
  if i.isalpha():
    alpcount += 1
  elif i.isalnum():
    numcount += 1
  elif i in "~!@#$%^&*':;<>,":
    specount += 1
print("letters count = ",alpcount)
print("Digits count = ",numcount)
print("Special cahrector count =",specount)
```

```python
# 9. Write a function to check if a given password is strong. A strong password should have at least 8
# characters, a mix of uppercase and lowercase letters, and at least one digit.
#  Input: "Password@123"  ,  Output: True
#  Input: "password123"   ,  Output: false
#  Input: "P@ssw0rd"      ,  Output: True

password = input("Enter a strong password :")
upper = False
lower = False
numeric = False

for i in password:
  if i.isupper():
    upper = True
  elif i.islower():
    lower = True
  elif i.isdigit():
    numeric = True

if len(password) >= 8 and upper and lower and numeric:
  print(True)
else:
  print(False)
```
