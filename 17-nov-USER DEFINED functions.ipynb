# **RECURSION**

```python
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n * factorial(n-1)
result = factorial(5)
print(result)

# factorial(5)    n == 5  else return 5*factorial(5-1) --- > returnil    5x24 return 120
# factorial(4)    n == 4  else return 4*factorial(4-1) --- > returnil    4x6  return 24
# factorial(3)    n == 3  else return 3*factorial(3-1) --- > returnil    3x2  return 6
# factorial(2)    n == 2  else return 2*factorial(2-1) --- > returnil    2x1  return 2
# factorial(1)    n == 1   if return 1  1*factorial(1-1) --- > returnil  1x1  return 1
# factorial(0)    n == 0   if return 1  1*factorial(0-1) --- > returnil  1x1  return 1
# n-1 conditions false, exited from function

# result = 5*4*3*2*1*1  = 120
```

```python
# pgm to find sum of natural numbers upto 10.

def sum1(num):
  if num == 0:
    return 0
  else:
    return num + sum1(num-1)

result = sum1(10)
print(result)
```

# **TYPES OF ARGUMENTS**

```python
# positional arguments

def student(name, age, place):
  print(name, age, place)

student("ameen", 27, "ekm")
# student("ameen", 27)   --> error, Ellaa arguments lekkum value passanam
```

```python
# default arguments

def student(name="no name", age=0, place="unknown"):
  print(name, age, place)
student("ameen", 27, "ekm")
student("ameen",27)
student("ameen")
student()
# student("ameen", 27, "ekm", "out of limit")  error,  position count should be equal
```

```python
# keyword argument

def student(name, age, place):
  print(name, age, place)
student(age = 20, place = "ekm", name = "ameen")

#count of positions equal avanam, but order of positions engane change ayalm no problem. Nammal key mention chythal mathi, Functionu ullilulla orderil printum
```

```python
# variable length arguments.
# *argument_name            ---> printing in TUPLE type

def addition(*argument_name):
  print(argument_name)
addition(10, 20, 30)
addition(54, 63, 87, 56, 12, 10)

print("\nvariety\n")

def addition(a,*argument_name):
  print(a,argument_name)
addition(10, 20, 30)
addition(54, 63, 87, 56, 12, 10)

print("\nvariety\n")

def addition(a,*argument_name):
  print(argument_name)
addition(10, 20, 30)
addition(54, 63, 87, 56, 12, 10)
```

```python
# kwargs
#  **   ------>  printing in dictionary type

def student(**argname):
  print(argname)

student(name="abc", age = 22, place = "ekm")
student(name = "abc", age = 20)
student(name = "abc")

print("\nvariety\n")

def student(**argname):
  print(argname)
  for i,j in argname.items():
    print(i,j)
student(name="abc", age = 22, place = "ekm")
student(name = "abc", age = 20)
student(name = "abc")
```

```python
# even or odd

result = lambda num: "even" if num%2 == 0 else "odd"
print(result(40))
print(result(55))
```

```python
def multiply(n):          # n=10
  return lambda a : n*a   # return lambda a:a*10
result = multiply(10)     # result = lambda a:a*10
print(result(2))          # lambda 2: 2*10
```
