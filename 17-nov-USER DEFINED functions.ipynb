```python
#   pgm to check whether a num is odd or even

def odd_or_even():
  num = int(input("Enter a number: "))
  if num % 2 == 0:
    print(num, "is even")
  else:
    print(num, "is odd")

odd_or_even()
```

```python
#   pgm to check a num is prime or not

def prime_check():
  num = int(input("Enter a number : "))
  count = 0

  for i in range(1, num+1):
    if num%i == 0:
      count += 1
  if count == 2:
    print(num, "is prime")
  else:
    print(num, "is not prime")

prime_check()
```

```python
#   to print factorial of a num

def factorial():
  num = int(input("Enter a number : "))
  result = 1
  for i in range(1, num+1):
    result *= i
  print(result)

factorial()
```

# **RECURSION**

```python
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n * factorial(n-1)
result = factorial(5)
print(result)

# factorial(5)    n == 5  else return 5*factorial(5-1) --- > returnil    5x24 return 120
# factorial(4)    n == 4  else return 4*factorial(4-1) --- > returnil    4x6  return 24
# factorial(3)    n == 3  else return 3*factorial(3-1) --- > returnil    3x2  return 6
# factorial(2)    n == 2  else return 2*factorial(2-1) --- > returnil    2x1  return 2
# factorial(1)    n == 1   if return 1  1*factorial(1-1) --- > returnil  1x1  return 1
# factorial(0)    n == 0   if return 1  1*factorial(0-1) --- > returnil  1x1  return 1
# n-1 conditions false, exited from function

# result = 5*4*3*2*1*1  = 120
```

```python
# pgm to find sum of natural numbers upto 10.

def sum1(num):
  if num == 0:
    return 0
  else:
    return num + sum1(num-1)

result = sum1(10)
print(result)
```

```python
# variable length arguments.
# *argument_name            ---> printing in TUPLE type

def addition(*argument_name):
  print(argument_name)
addition(10, 20, 30)
addition(54, 63, 87, 56, 12, 10)

print("\nvariety\n")

def addition(a,*argument_name):
  print(a,argument_name)
addition(10, 20, 30)
addition(54, 63, 87, 56, 12, 10)

print("\nvariety\n")

def addition(a,*argument_name):
  print(argument_name)
addition(10, 20, 30)
addition(54, 63, 87, 56, 12, 10)
```

```python
# kwargs
#  **   ------>  printing in dictionary type

def student(**argname):
  print(argname)

student(name="abc", age = 22, place = "ekm")
student(name = "abc", age = 20)
student(name = "abc")

print("\nvariety\n")

def student(**argname):
  print(argname)
  for i,j in argname.items():
    print(i,j)
student(name="abc", age = 22, place = "ekm")
student(name = "abc", age = 20)
student(name = "abc")
```

```python
def multiply(n):          # n=10
  return lambda a : n*a   # return lambda a:a*10
result = multiply(10)     # result = lambda a:a*10
print(result(2))          # lambda 2: 2*10
```

```python
# filter()            -----> filtering only true or false condition satisfied elements or charecters
#filter(funtion, iterable)

# num = [45, 78, 12, 90, 221]
  # def function():
  #   code
# filter(isodd, num)



num = [45, 78, 12, 90, 221]
def isodd(n):
  return n%2 == 1
result = filter(isodd, num)
print(list(result))


# OR

result = filter(lambda n:n%2==1, num)
print(list(result))
```

```python
# map(function, iterable)    ------------> works on every element or characters

num = [45, 78, 12, 90, 221]
def double(n):
  return n*2
result = map(double, num)
print(list(result))

#OR


result = map(lambda n:n*2, num)
print(list(result))
```

```python
# reduce()  -----------> returns only a single value

#reduce(function, iterable)

from functools import reduce
num = [1, 2, 3, 4]
def getsum(x, y):
  return x+y
result = reduce(getsum, num)
print(result)

# x=1 y=2 return 1+2 3
# x=3 y=3 return 3+3 6
# x=6 y=4 return 6+4 10
```

```python
def demo():
  name = "anu"#-> local variable
  print(name)
demo()
#print(name) #-> we can't acces it out side of the function , only allowed any alterations or calling inside the function
```

```python
#  Create a number guessing game where the computer selects a random number between
#  1 and 100 and the user has to guess it. Provide hints like "too high" or "too low"
#  based on user's input. Also print the number of attempts

import random

system = random.randrange(1 , 101)
count = 1
while True:
  user = int(input("Guess a number: "))
  if user == system:
    print("Success")
    print("You attempted ", count, "Times")
    break
  elif user > system:
    print("Your guess too high")
    count += 1
  elif user < system:
    print("Your guess too low")
    count += 1
```

```python
from datetime import timedelta, datetime   # timedelta is used to fimd future or past

now = datetime.now()
print("Current date and time ", now)
future_date = now + timedelta(days = 5)
print("Date after 5 days =", future_date)
past_time = now - timedelta(hours = 2)
print("Time before 2 hours was ", past_time)
```
