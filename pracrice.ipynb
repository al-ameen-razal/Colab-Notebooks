```python
# print palindrome numbers from a list

list1 = [122, 123, 12321, 121, 145]

for i in list1:
  out = 0
  number = i

  while number > 0:
    lastdigit = number%10
    out = out*10 + lastdigit
    number = number//10

  if out == i:
    print(i)
```

```python

listmakelambda = lambda l,k,step :[i for i in range(l,k+1,step)]

print(listmakelambda(10,15,3))

```

```python
is_even = lambda n: f"{n} is even" if n%2 == 0 else f"{n} is odd"

print(is_even(2))
print(is_even(1))
```
