```python
# print palindrome numbers from a list

list1 = [122, 123, 12321, 121, 145]

for i in list1:
  out = 0
  number = i

  while number > 0:
    lastdigit = number%10
    out = out*10 + lastdigit
    number = number//10

  if out == i:
    print(i)
```
