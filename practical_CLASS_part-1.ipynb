1. Define a class Car with 2 class varibles brand and color. Print their values using the class name.

```python
class Car:
  def __init__(self, brand, color):
    self.brand_name = brand
    self.color = color

  def show(self):
    print(f"Your car brand is {self.brand_name}")
    print(f"Your car's color is {self.color}")

car1 = Car("Toyota", "White")
car1.show()
```

2. Create a class Laptop with class variables brand and price. Access and print these values using an object of class.

```python
class Laptop:
  def __init__(self, brand, price):
    self.brand = brand
    self.price = price

  def show(self):
    print(f"Price for {self.brand} laptop is {self.price} Rs./-")

lap1 = Laptop("HP", 33400)
lap1.show()

lap2 = Laptop("Apple", 98500)
lap2.show()
```
