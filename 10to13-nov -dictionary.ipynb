```python
# Program to manage a shop's inventory.
# Implement the following functionalities using a dictionary.
# Continue this process until the user enters 'exit'
# 1, Add a new product
# 2, Update the quantity of existing functions
# 3, Display the inventory

shop = {
    "pen":12,
    "pencil":7
}

while True:
  # Changed message to be more readable
  message = "Enter an option\n"
  message += "1) 'Add' Add new product\n"
  message += "2) 'Display' Display inventory\n"
  message += "3) 'Modify' Modify the qty of exixsting produdct\n"
  message += "4) 'Exit' Exit\n"
  user = input(message).lower()

  # Changed order of if conditions to meet with the order of display message
  # Makes it easy to modify in future
  if user == "add" or user == "1":  # Added choosing number as option
    product = input("Enter product name: ")
    qty = int(input("Enter the product quantity: "))
    shop[product] = qty
    print("The stock update is")
    print(shop)
    print()

  elif user in ["display", "d", "2"]: # also we can use this, makes the code simple
    # but this might not be optimal since using list (not a simple datatype)
    # require more computation than using == operator in simple datatypes
    # so use it with BODHAM
    print("Your inventory is")
    # Changed items to print on below - makes it more readable
    print(shop)
    print()

  elif user == "modify" or user == "3":
    product = input("Enter product name: ")
    if product not in shop:
      print(product.upper(), "is not found in your inventory. Please add as new product.")
      dicision = input("Do you want to add this new product? (Yes/No)")
      if dicision == "yes":
        qty = int(input("Enter the product quantity: "))
        shop[product] = qty
        print("The stock update is", shop)
        print()
      else:
        print()
        continue
    else:
      qty = int(input("Enter the product quantity: "))
      shop[product] += qty
      print()
    print("The stock update is")
    print(shop)

  elif user == "exit" or user == "4":
    print("See you again Admin.")
    break

  else:
    print("You can't do '",user.upper(),"' in my inventory. Your present Inventory is")
    print(shop,"\n")
    print("Please choose one of the given options!!!")
```

COPY

```python
# SHALLOW COPY  - the changes in copy also effect the original dictionary

student = {
    "name":"alameen",
    "marks":{"maths":78, "english":45}
}
print(student)
dict1 = student.copy()
print(student,dict1)
dict1["marks"]["maths"] = 2
print(student,dict1)

```

```python
# deep copy  -- changes in copy  doesnt effect the original dictionary

import copy
student = {
    "name":"alameen",
    "marks":{"maths":78, "english":45}
}
print(student)
dict1 = copy.deepcopy(student)
print(student,dict1)
dict1["marks"]["maths"] = 2
print(student,dict1)

```
