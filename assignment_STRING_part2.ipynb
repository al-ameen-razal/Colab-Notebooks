```python
# 14. Write a program to reverse each character of a string using while loop.

sentence = input("Enter a sentence : ")
lastchar = len(sentence) - 1
result = ""

while lastchar >= 0:
  result += sentence[lastchar]
  lastchar -= 1

print(result)
```

```python
# 20. Write a program that takes two strings as input and finds the common characters between them.

word1 = input("Enter the first sentence : ")
word2 = input("Enter the second sentence : ")
set1 = set(word1)
set2 = set(word2)
print("Common charector : ",set1&set2)

# OR

cache = ""
shorter = (min(word1,word2))
larger = max(word1,word2)
print("\nCommon characters:- ")
for i in shorter:
  if i in larger:
    if i not in cache:
      print(i,end=",")
      cache += i
```

```python
# 1. Write a Python program that asks the user to enter a string and checks if it's an isogram.
#     An isogram is a word or phrase without any repeating letters.

text = input("Type the word here: ")
testing = text.lower().replace(" ", "")
flag = True

for i in range(len(testing)):
    for j in range(i + 1, len(testing)):
        if testing[i] == testing[j]:
            flag = False
            break

if flag:
    print("YES,'",text,"' is an isogram.")
else:
    print("NO,'",text,"' is not an isogram.")

```
