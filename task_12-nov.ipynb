```python
#2.Remove all punctuation from a string using list.

sentence = "Hello, world! How's the day. Don't you feel good?"
clean_chars = []

for i in sentence:
    if i.isalnum() or i == " ":
        clean_chars.append(i)
clean_sentence = ''.join(clean_chars)

print(clean_sentence)
```

```python
#3. Remove all repeated words from a sentence.

sentence = "I'm a robot, you are a robot, he is a robot, we are the robots"
sentence_list = sentence.split()
result = list()

for i in sentence_list:
  if i not in result:
    result.append(i)
    out = " ".join(result)
print(out)
```

```python
#4. Rearrange characters: vowels first, consonants next.

sentence = "ajilemuvo"
list1 = []
list2 = []
for i in sentence:
  if i in "aeiou":
    list1.append(i)
  else:
    list2.append(i)
list3 = list1+list2
result = " ".join(list3)
print(result)
```

```python
#5. Remove all words shorter than 4 letters

sentence = "I'm a robot you are a robot he is a robot, we are the robots"
list1 = sentence.split()
list2 = []
for i in list1:
  length = len(i)
  if (length) >= 4:
    list2.append(i)
result = " ".join(list2)
print(result)
```

```python
#8. Replace every second vowel with ‘#’.


input_text = "Beautiful day in the neighbourhood"
vowel = 'aeiouAEIOU'
count = 0

for word in input_text:
  for letter in word:
    if letter in vowel:
      count += 1
      if count%2 == 0:
        letter = "#"
    print(letter,end="")
```

```python
# @title
#10.  Encode and decode words by reversing vowels only
#     text = "education"
#     Output: odicatuen

vowel = 'aeiou'
word = "education"
vowlist = []
conlist = []
concat = []

for letter in word:
  if letter in vowel:
    vowlist.append(letter)
  else:
    conlist.append(letter)
print(vowlist[::-1], conlist)

length = len(vowlist+conlist)
print(length)

for i in range(len(vowlist)):
  concat.append(vowlist[i])
  for j in range(len(conlist)):
    concat.append(conlist[i])
    break
print(concat)
```

```python
#10.  Encode and decode words by reversing vowels only
#     text = "education"
#     Output: odicatuen


vowels = 'aeiou'
word = "education"

# Extract and reverse vowels
vowel_list = [ch for ch in word if ch in vowels]
vowel_list.reverse()

# Build new word, replacing vowels with reversed ones
result = ''
vowel_index = 0

for ch in word:
    # print(ch)
    if ch in vowels:
        result += vowel_list[vowel_index]
        vowel_index += 1
    else:
        result += ch

print(result)
```

```python
# 13.Replace spaces with next character.

string = "python makes coding fun"
items = string.split()
new_items = [items[0]]

for item in items[1:]:
  add_items = item[0] + item
  new_items.append(add_items)
print("".join(new_items))
```

```python
# 14.Rearrange words in descending order of vowel count.

text = "programming is beatiful and fun"
list1 = text.split()
result = []
output = []
for i in list1:
  count = 0
  for j in i:
    if j in "aeiou":
      count += 1

  # print(count)
  result.append(str(count) + i)
  # print(result)
result.sort(reverse=True)
#print(result)

for word in result:
  output.append(word[1:])
#print(output)
print(" ".join(output))
```

```python
# 15.Remove duplicate letters but keep last occurrence.

text = "programming"
rev = text[::-1]
#print(rev)
result = []

for i in rev:
  if i not in result:
    result.append(i)
#print(result[::-1])
print("".join(result[::-1]))
```
