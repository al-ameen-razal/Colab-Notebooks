```python
# Program to find divide a number upto given decimal points.

# 3 inputs - firsr number, division number, precision
# 1 ouput  - number after division with precision

# Example
# Input - 22, 7, 5
# Output - 3.14287

# Input -10, 3, 30
# Output - 3.333333333333333333333333333333



a = int(input("Enter the number : "))
b = int(input("Enter the divider : "))
c = int(input("Enter the precision : "))

integer = a//b
floating = a%b
result = str(integer) + "."
i = 0
while i < c:
  repeat = floating * 10
  integer = repeat//b
  result += str(integer)
  floating = repeat%b
  i += 1

print(result)
print(len(result))
```

```python
# Task:
# Ask the user to enter the distance of a trip in miles,
# and the fuel efficiency of their car in miles per gallon (MPG).
# Then calculate the total fuel cost of the trip in dollars.
# You must write at least three user-defined functions:

# 1. A function to convert miles to kilometers
# 2. A function to convert MPG to liters-per-100-km (1 mpg is equal to 235.2 liters per 100 km.)
# 3. A function to calculate the total fuel cost, using:
# distance in km
# L/100km
# current fuel price per liter = 2 dollars


def kilometer(miles):
  mpg_to_km =  miles * 1.609344
  return mpg_to_km

def l_per_100km(mileage):
  rel_per_100 =  235.215/mileage
  return rel_per_100

def fuel_cost(distance):
  fuel_cost = kilometer(distance) * fuel_price
  return fuel_cost

distance = float(input("Enter the total diatance (in miles): "))
mileage = float(input("Enter your CAR's mleage in MPG : "))
fuel_price = float(input("Enter the Fuel price per litter : "))

print()
print("Distance =",kilometer(distance),"KM")
print("Mileage = ",l_per_100km(mileage), "L/100km")
print("Total fuel cost = ",fuel_cost(fuel_price),"Dollars")
```

```python
# negative or positive

def neg_pos(num):
  if num > 0:
    return "positive"
  if num < 0:
    return "negative"
  return "Zero"

neg_pos(int(input()))
```

```python
class calculator:
  a = 10
  b = 5

  def addition(self):
    return self.a + self.b

  def substraction(self):
    return self.a - self.b

  def change_value(self,new_a, new_b):
    self.a, self.b = new_a, new_b
    return self.a + self.b


operation1 = calculator()
print(operation1.addition())
print(operation1.substraction())
print(operation1.change_value(100, 60))
```

```python
class Calculator:
  def __init__(self, num1, num2):
    self.a = num1
    self.b = num2

  def add(self):
    print(f"Sum = {self.a + self.b}")

  def sub(self):
    print(f"Difference = {self.a - self.b}")

  def change(self, num1, num2):
    self.a = num1
    self.b = num2

first = Calculator(1, 2)
first.add()
first.sub()
first.change(10, 5)
first.add()
first.sub()
```
