```python
# Program to find divide a number upto given decimal points.

# 3 inputs - firsr number, division number, precision
# 1 ouput  - number after division with precision

# Example
# Input - 22, 7, 5
# Output - 3.14287

# Input -10, 3, 30
# Output - 3.333333333333333333333333333333



a = int(input("Enter the number : "))
b = int(input("Enter the divider : "))
c = int(input("Enter the precision : "))

integer = a//b
floating = a%b
result = str(integer) + "."
i = 0
while i < c:
  repeat = floating * 10
  integer = repeat//b
  result += str(integer)
  floating = repeat%b
  i += 1

print(result)
print(len(result))
```

```python
# odd or even

def oddOReven(num):
  if num%2 == 0:
    return "even"
  return "odd"

oddOReven(int(input("Enter a num: ")))
```

```python
# negative or positive

def neg_pos(num):
  if num > 0:
    return "positive"
  if num < 0:
    return "negative"
  return "Zero"

neg_pos(int(input()))
```
