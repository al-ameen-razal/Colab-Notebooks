```python
# Program to find divide a number upto given decimal points.

# 3 inputs - firsr number, division number, precision
# 1 ouput  - number after division with precision

# Example
# Input - 22, 7, 5
# Output - 3.14287

# Input -10, 3, 30
# Output - 3.333333333333333333333333333333



a = int(input("Enter the number : "))
b = int(input("Enter the divider : "))
c = int(input("Enter the precision : "))

integer = a//b
floating = a%b
result = str(integer) + "."
i = 0
while i < c:
  repeat = floating * 10
  integer = repeat//b
  result += str(integer)
  floating = repeat%b
  i += 1

print(result)
print(len(result))
```

```python
# Task:
# Ask the user to enter the distance of a trip in miles, and the fuel efficiency of their car in miles per gallon (MPG).
# Then calculate the total fuel cost of the trip in dollars.
# You must write at least three user-defined functions:

# 1. A function to convert miles to kilometers
# 2. A function to convert MPG to liters-per-100-km (1 mpg is equal to 235.2 liters per 100 km.)
# 3. A function to calculate the total fuel cost, using:
# distance in km
# L/100km
# current fuel price per liter = 2 dollars


def kilometer(miles):
  mpg_to_km =  miles * 1.609344
  return mpg_to_km

def l_per_100km(mileage):
  rel_per_100 =  235.215/mileage
  return rel_per_100

def fuel_cost(km):
  fuel_cost = kilometer(km) * 2
  return fuel_cost
  # used 2 dollars as the current fuel price



while True:
  print("\n")
  message = "Select the option : \n"
  message += "1. Conver travelled dist. from miles to KM : \n"
  message += "2. Convert mileage to Litter/100km \n"
  message += "3. Find my total fuel cost \n"
  message += "4. Exit\n"
  user = input(message)
  if user in ("4", "exit"):
    print("Exited")
    break

  if user in ["1", "km", "dist"]:
    result =  kilometer(int(input("Enter distance in Miles to convert into KM : ")))
    print(result,"KM")

  if user in ["2", "mileage", "litter/100km"]:
    result = l_per_100km(float(input("Enter the mileage of your car in MPG : ")))
    print(result,"Litter/100KM")

  if user in ["3", "cost", "expense"]:
    result = fuel_cost(int(input("Enter distance in Miles : ")))
    print(result,"Dollars")
```
