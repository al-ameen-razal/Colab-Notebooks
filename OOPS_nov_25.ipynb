```python
class Student:
  def getStudentData(self, name, rollno):
    self.name = name
    self.rollno = rollno

  def displayStudentData(self):
    print("Name : ", self.name)
    print("Rollno : ", self.rollno)

class Marks(Student):
  def getMarks(self, m1, m2):
    self.mark1 = m1
    self.mark2 = m2

  def displayMarks(self):
    print("Mark1 = ", self.mark1)
    print("Mark2 = ", self.mark2)

  def calculate(self):
    self.total = self.mark1 + self.mark2
    self.p = self.total / 200*100
    print("Total = ", self.total)
    print("Percentage = ", self.p)

class Grade(Marks):
  def getGrade(self):
    if self.p >= 90:
      print("A grade")
    elif self.p >= 70:
      print("B grade")
    elif self.p >= 50:
      print("C grade")
    elif self.p >= 40:
      print("D grade")
    else:
      print("Failed")

  def display(self, name, rollno, m1, m2):
    self.getStudentData(name, rollno)
    self.getMarks(m1, m2)
    self.displayStudentData()
    self.displayMarks()
    self.calculate()
    self.getGrade()
    print()

s1 = Grade()
s1.display("Ameen", 11, 10, 10)

s2 = Grade()
s2.display("Hari", 88, 95, 99)
```
