1. Write a Python program that finds and prints all pairs of elements in a list that sum up to a specific target value

```python
numbers = [1, 2, 3, 4, 5, 6 ,7]
target = int(input("Enter the target SUM : "))

for i in range(len(numbers)):
  for j in range(i+1,len(numbers)):
    if numbers[i] + numbers[j] == target:
      print(f"{numbers[i]} + {numbers[j]} = {target}")
```

3. Write a Python program that presents a menu to the user for performing various operations on a list
(e.g. append, remove, sort, reverse) and updates the list based on the user's choice. Implement the menu using a loop.

```python

bag = ["a", "d", "c", "b", "f", "e"]
while True:
  user = input("1. Add element\n2. Remove element\n3. Sort the elements\n4. Reverse the list\n5. Exit\n")

  if user in ["5", "exit"]:
    break
  elif user in ["1", "add", "append"]:
    element = input("Enter the element to add to your bag : ")
    bag.append(element)
    print(element, "added\nYour bag -->", bag)
    print()
  elif user in ["2", "remove", "delete"]:
    element = input("Enter the element to be removed : ")
    if element not in bag:
      print(element, "is not your item")
      print("Your bag -->", bag)
      print()
    else:
      bag.remove(element)
      print(element, "removed\nYour bag -->", bag)
      print()
  elif user in["3", "sort", "arrange"]:
    order = input("1) Ascending\n2) Descenting\n")
    if order in ["1", "ascending"]:
      print(sorted(bag))
    elif order in ["2", "descending"]:
      print(sorted(bag, reverse=True))
    print()
  elif user in ["4", "reverse", "up side down"]:
    rev = bag[::-1]
    print(f"Your bag --> {rev}")
    print()
  else:
    print("Invalid entry")
    print()
```

lists [1, 2, 3] and [3, 4, 5], the output should be [3].

```python
list1 = [1, 2, 2, 3]
list2 = [3, 2, 4, 5]
out = []

for i in list1:
  if i in list2 and i not in out:
    out.append(i)
for j in list2:
  if j in list1 and j not in out:
    out.append(j)

print(out)
```

7. Rearrange the list so that positive and negative numbers appear alternately.


```python
list1 = [1, 2, 3, -4, -5, -6]
positive = []
negative = []
result = []

for i in list1:
  if i >= 0:
    positive.append(i)
  else:
    negative.append(i)

k = max(len(negative),len(positive))

for j in range(k):
  result.extend([positive[j], negative[j]])

print(result)
```
